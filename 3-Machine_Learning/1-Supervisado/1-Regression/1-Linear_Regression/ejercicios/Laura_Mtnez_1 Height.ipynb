### ¿Es una buena técnica para este tipo de problemas? Demuéstralo antes de entrenar un modelo con un análisis descriptivo.

```python
df = pd.DataFrame(lista_alumnos, columns=['Nombre', 'Edad', 'Altura'])
df.head()
```

```python
sns.pairplot(df);
```

### Entrena un modelo de regresión lineal

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
X = df[['Edad']]
X
```

```python
y = df[['Altura']]
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 4, random_state= 42)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.coef_
```

```python
X_test
```

```python
lr.predict(X_test)
```

### ¿Qué error se comete? Calcula los errores que está cometiendo tu modelo de forma manual con python. Aparte, usa las métricas de sklearn MAE, MAPE, MSE y el RMSE.

```python
df_2 = df
```

```python
df_2['prediccion'] = lr.predict(df_2[['Edad']])
```

```python
#En Python, ravel() es una función del módulo NumPy (o un método de los arrays) que sirve para aplanar un array multidimensional (como una matriz de 2D o 3D) convirtiéndolo en un array unidimensional (de una sola dimensión).
df_2['error'] = df_2['Altura'] - lr.predict(df_2[['Edad']]).ravel()
df_2['error cuad'] = df_2['error']**2
df_2
```

```python
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, mean_squared_error, root_mean_squared_error, r2_score
```

```python
y_pred = lr.predict(df_2[['Edad']])
```

```python
mean_absolute_error(df_2[['Altura']], y_pred)   #Este es el error en metros, ed, de media me confundo 6cm
```

```python
mean_absolute_percentage_error(df_2[['Altura']], y_pred)   #Este es el error en porcentaje
```

```python
mean_squared_error(df_2[['Altura']], y_pred)   #OJO unidades al cuadrado, penaliza más los errores grandes
```

```python
root_mean_squared_error(df_2[['Altura']], y_pred)   #Aqui sí está en las uds, porque hace la raíz cuadrada
```

```python
r2_score(df_2[['Altura']], y_pred)
```
