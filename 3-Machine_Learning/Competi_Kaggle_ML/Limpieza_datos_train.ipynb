```python
# 1. Extraer el ancho y el alto usando expresiones regulares
resolucion = df['ScreenResolution'].str.extract(r'(\d+)x(\d+)')
df['Res_Ancho'] = resolucion[0].astype(int)
df['Res_Alto'] = resolucion[1].astype(int)

# 2. Crear columnas binarias (1 si contiene la palabra, 0 si no)
df['Touchscreen'] = df['ScreenResolution'].str.contains('Touchscreen', case=False).astype(int)
df['IPS'] = df['ScreenResolution'].str.contains('IPS', case=False).astype(int)

# 3. (Opcional) Calcular PPI si tienes la columna 'Inches' como número
df['PPI'] = ((df['Res_Ancho']**2 + df['Res_Alto']**2)**0.5) / df['Inches']

# 4. Eliminar la columna original de texto
df = df.drop(columns=['ScreenResolution'])
```

```python
# 1. Extraer la frecuencia en GHz como número decimal
df['Cpu_GHz'] = df['Cpu'].str.extract(r'(\d+(?:\.\d+)?)GHz').astype(float)

# 2. (Opcional) Detectar si es procesador Intel de alto rendimiento (HQ, HK, H)
df['Cpu_Alto_Rendimiento'] = df['Cpu'].str.contains(r'\d+(?:HQ|HK|H)\b|Ryzen').astype(int)
# 3. Agrupar la familia del procesador
def categorizar_cpu(texto):
    if 'Intel Core i7' in texto:
        return 'Intel Core i7'
    elif 'Intel Core i5' in texto:
        return 'Intel Core i5'
    elif 'Intel Core i3' in texto:
        return 'Intel Core i3'
    elif 'Intel' in texto:
        return 'Otro Intel'
    elif 'AMD' in texto:
        return 'AMD'
    else:
        return 'Otro'

df['Cpu_Gama'] = df['Cpu'].apply(categorizar_cpu)

# 4. Eliminar la columna original
df = df.drop(columns=['Cpu'])
```

```python
import re

def extraer_almacenamiento(texto):
    # Inicializamos los contadores en GB (unificando Flash Storage dentro de SSD)
    capacidades = {'SSD': 0, 'HDD': 0, 'Hybrid': 0}
    
    # Separamos por '+' si hay dos discos
    partes = str(texto).split('+')
    
    for parte in partes:
        parte = parte.strip()
        # Buscamos el número (incluyendo decimales como 1.0) y la unidad (GB o TB)
        match = re.search(r'(\d+(?:\.\d+)?)\s*(GB|TB)', parte)
        if match:
            valor = float(match.group(1))
            unidad = match.group(2)
            
            # Convertimos TB a GB (usando 1000 o 1024)
            if unidad == 'TB':
                valor *= 1000
                
            # Sumamos al tipo de disco correspondiente (Flash Storage va a SSD)
            if 'SSD' in parte or 'Flash Storage' in parte:
                capacidades['SSD'] += int(valor)
            elif 'HDD' in parte:
                capacidades['HDD'] += int(valor)
            elif 'Hybrid' in parte:
                capacidades['Hybrid'] += int(valor)
                    
    return pd.Series([
        capacidades['SSD'], 
        capacidades['HDD'], 
        capacidades['Hybrid']
    ])

# Aplicamos la función al DataFrame (solo se crean 3 columnas)
df[['SSD_GB', 'HDD_GB', 'Hybrid_GB']] = df['Memory'].apply(extraer_almacenamiento)

# Eliminamos la columna original de texto
df = df.drop(columns=['Memory'])
```

```python
import pandas as pd

# 1. Extraer la marca principal (primera palabra: Intel, Nvidia, AMD)
df['Gpu_Marca'] = df['Gpu'].apply(lambda x: x.split()[0])

# 2. Clasificar por gama de rendimiento
def categorizar_gpu(texto):
    if 'Intel' in texto:
        return 'Intel Integrada'
    elif 'GTX' in texto or 'RTX' in texto:
        return 'Nvidia Gaming (GTX/RTX)'
    elif 'Quadro' in texto:
        return 'Nvidia Profesional (Quadro)'
    elif 'Nvidia' in texto:
        return 'Nvidia Básica (MX/900M)'
    elif 'Pro' in texto or 'FirePro' in texto:
        return 'AMD Profesional'
    else:
        return 'AMD Estándar'

df['Gpu_Gama'] = df['Gpu'].apply(categorizar_gpu)

# 3. Eliminar la columna original
df = df.drop(columns=['Gpu'])
```

```python
def limpiar_opsys(os):
    if 'Windows 10' in os:
        return 'Windows 10'
    elif 'Mac' in os or 'macOS' in os:
        return 'macOS'
    elif os in ['Windows 7', 'Linux', 'Chrome OS', 'No OS']:
        return os
    else:
        return 'Otro'

df['OpSys'] = df['OpSys'].apply(limpiar_opsys)
```
