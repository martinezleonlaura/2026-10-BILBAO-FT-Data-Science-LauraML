## Ejercicio 1
Dada la siguiente lista:
> ```ejer_1 = [1,2,3,4,5]```

> ```ejer_1 = [5,4,3,2,1]```

````python
ejer_1 = [1,2,3,4,5]
ejer_1.sort()
ejer_1.reverse()        #OJO! ".reverse()"" solo da la vuelta vuelta a la lista, no ordena de mayor a menor por eso he hecho el sort() antes aunque en este caso ya estban ordenados
print(ejer_1)
````

Otra forma

````python
ejer_1 = [1,2,3,4,5]
ejer_1 = ejer_1[::-1]       #Sobreescribo la lista recorriéndola al revés con slicing
print(ejer_1)
````

## Ejercicio 3
Crea una lista nueva con todas las combinaciones de las siguientes dos listas:
> ```ejer_3_1 = ["Hola", "amigo"]```
>
> ```ejer_3_2 = ["Que", "tal"]```

Obten el siguiente output:

```['Hola Que', 'Hola tal', 'amigo Que', 'amigo tal']```

````python
ejer_3_1 = ["Hola", "amigo"]
ejer_3_2 = ["Que", "tal"]
ejer_3_combo = []

for x in ejer_3_1:
    for y in ejer_3_2:
        ejer_3_combo.append((x, y))

print(ejer_3_combo)
````

## Ejercicio 4
Dada la siguiente lista, encuentra por el índice el valor 45 y sustitúyelo por el 0
> ```ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]```

````python
ejer_4 = [20, 47, 19, 29, 45, 67, 78, 90]

for num in ejer_4:
    if num == 45:
        indice = ejer_4.index(num)
        ejer_4[indice] = 0
print(ejer_4)
````

## Ejercicio 5
Dada la siguiente lista, elimina todos los valores iguales a 3
> ```ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]```

TIP: No intentes eliminar elementos sobre la lista que estás iterando. Haz una copia con `ejer_5.copy()`.

````python
ejer_5 = [3, 20, 3, 47, 19, 3, 29, 45, 67, 78, 90, 3, 3]
copia = ejer_5.copy()

for num in copia:
    if num == 3:
        indice = ejer_5.index(num)
        ejer_5.pop(indice)
print(ejer_5)
````

## Ejercicio 11
Convierte la siguiente lista en un diccionario
> ```ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]```

TIP: Puedes usar dos variables de control en un bucle `for`. Una por cada valor de las tuplas.

````python
ejer_11 = [("x", 1), ("x", 2), ("x", 3), ("y", 1), ("y", 2), ("z", 1)]            #Lista de tuplas
diccionario = {}

for clave, valor in ejer_11:            #Recorro la lista. La primera pos de cada tupla (las letras) será una clave, y la segunda (los números) serán los valores
    if clave not in diccionario:        #Primero compruebo que no se repitan las clave
        diccionario[clave] = []         #Creo una lista para las distintas claves, donde iré almacenando los valores
    diccionario[clave].append(int(valor))       

print(diccionario)
````

## Ejercicio 12
1. Crea una lista ordenada ascendente con las claves del diccionario
2. Crea otra lista ordenada descendente con los valores

5. Itera la clave y el valor del diccionario con un único for

## Ejercicio 14
Suma todos los valores del dicionario
> ```ejer_14 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}```

````python
ejer_14 = {1: 11, 2: 22, 3: 33, 4: 44, 5: 55}

lista_valores_ejer_14 = list(ejer_14.values())
#print(lista_valores_ejer_14)

suma_valores = 0
for num in lista_valores_ejer_14:
    suma_valores = suma_valores + num

print(f"La suma de todos los valores del diccionario 'ejer_14' es de: {suma_valores}")
````
