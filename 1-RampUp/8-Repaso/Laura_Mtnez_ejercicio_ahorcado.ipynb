## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
def ahorcado():
    
    letras_acertadas = []
    letras_falladas = []
    contador = 0
    intentos = 6

    palabra = input(f"Introduce la palabra que tendrá que adivinar tu adversario:").upper()

    tablero = ["_"] * len(palabra)
    print(" " .join(tablero))
    print()

    while contador < intentos:

        letra_jugada = input("Introduce una letra:").upper()

        if len(letra_jugada) > 1:
            print(f"Por favor, introduce una sola letra")
            continue

        if letra_jugada in letras_acertadas or letra_jugada in letras_falladas:
            print(f"Ya has intentado esa letra. Por favor, introduce otra letra")
            continue

        if letra_jugada in palabra:
            for i,letra in enumerate(palabra):
                if letra == letra_jugada:
                    tablero[i] = letra_jugada
            letras_acertadas.append(letra_jugada)
        else:
            contador = contador + 1
            letras_falladas.append(letra_jugada)
            print(f"La letra no pertenece a la palabra. Te quedan {intentos - contador} intentos.")

        print(" ".join(tablero))
        print(f"Letras falladas: {letras_falladas}")
        print()
            
        if "_" not in tablero:
            print("YOU WIN!!!.")
            break
                
        if contador == intentos:
            print(f"GAME OVER... La palabra era: {palabra}")

````

````python
ahorcado()
````
