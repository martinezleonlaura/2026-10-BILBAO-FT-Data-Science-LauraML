## ¿Cómo funciona el juego?
Vamos a realizar una versión que tiene algunas particularidades respecto al juego original, de manera que sea más sencillo el desarrollo. Veamos cómo funciona:

1. Hay dos jugadores: tú y la máquina
2. Un **tablero de 10 x 10** posiciones donde irán los barcos
3. Lo primero que se hace es colocar los barcos. Para este juego **los barcos se colocan de manera aleatoria. Ahora bien, puedes empezar colocando los barcos en unas posiciones fijas, que no cambien con cada partida, y después implementarlo aleatoriamente, ya que es más complejo. Los barcos son:**
    * 4 barcos de 1 posición de eslora
    * 3 barcos de 2 posiciones de eslora
    * 2 barcos de 3 posiciones de eslora
    * 1 barco de 4 posiciones de eslora

4. Tanto tú, como la máquina tenéis un tablero con barcos, y se trata de ir "disparando" y hundiendo los del adversario hasta que un jugador se queda sin barcos, y por tanto, pierde.
5. Funciona por turnos y empiezas tú.
6. En cada turno disparas a una coordenada (X, Y) del tablero adversario. **Si aciertas, te vuelve a tocar**. En caso contrario, le toca a la máquina.
7. En los turnos de la máquina, si acerta también le vuelve a tocar. ¿Dónde dispara la maquina? A un punto aleatorio en tu tablero.
8. Si se hunden todos los barcos de un jugador, el juego acaba y gana el otro.

        * Adicionalmente la clase tablero necesitará otra lista de listas, ¿por qué? porque el tablero de la máquina tendrá internamente una lista de listas con sus barcos (lo que no vemos) y hará falta otra lista de listas (que sí veremos nosotros) con los disparos efectuados, para saber dónde tenemos que disparar.
    * **Inicializar el tablero**, es decir, colocar los barcos. Puedes pasar por alto el hecho de que tengan que tener espacios entre ellos pero si los colocas aleatoriamente, mucho cuidado aquí de poner los barcos dentro del tablero, y de no colocar unos barcos encima de otros :)

3. Una vez ya tienes modelizado tu tablero, hay que montar el programa que se ejecutara desde un **main.py**:

```python
DIMENSION = 10

SIMBOLO_AGUA = " "

SIMBOLO_BARCO = "B"

SIMBOLO_TOCADO = "x"

SIMBOLO_FALLO = "o"

BARCOS = {"Ballenero" : 4,
          "Trainera_1" : 3,
          "Trainera_2" : 3,
          "Txalupa_1" : 2,
          "Txalupa_2" : 2,
          "Txalupa_3" : 2,
          "Batel_1" : 1,
          "Batel_2" : 1,
          "Batel_3" : 1,
          "Batel_4" : 1}

import random
import pprint

class Tablero:

    #Atributos COMUNES
    dimension = DIMENSION

    #Atributos PROPIOS de cada Tablero
    def __init__(self, jugador:str, barcos: dict = BARCOS):
        self.jugador = jugador
        self.barcos = barcos
        #Cada tablero de cada jugador se compone de dos tableros: uno donde posiciona los barcos, y otro donde apunta los disparos
        self.tablero_barcos = [[SIMBOLO_AGUA for i in range(self.dimension)] for j in range(self.dimension)]
        self.tablero_disparos = [[SIMBOLO_AGUA for i in range(self.dimension)] for j in range(self.dimension)]
        self.coordenadas_barco = {}     #Para comprobar luego si he hundido el barco

    #Diferentes MÉTODOS para la clase Tablero
    def colocar_barcos_random(self):
        '''
        Método para elegir las coordenadas los barcos de manera aletoria y colocarlos.
        Dentro de este método llamo a otros métodos:
            1.- Para validar que la posición es corecta (no se sale del tablero, ni se solapa con otros barcos).
            2.- Una vez validada la posición, colocar el barco propiamente dicho.
        '''
        for nombre, eslora in self.barcos.items():
            colocado = False
            while not colocado:
                orientacion = random.choice(["N-S", "E-O"])
                x = random.randint(0, self.dimension - 1)
                y = random.randint(0, self.dimension - 1)
                if self.validar_posicion(orientacion, x, y, eslora):
                    self.colocar_barco_en_tablero(orientacion, x, y, eslora)
                    coordenadas_barco = []     #Guardo las coordenadas del barco para comprobar luego si lo he hundido
                    if orientacion == "E-O":
                        for i in range (y, y + eslora):
                            coordenadas_barco.append((x, i))
                    else:
                        for i in range (x, x + eslora):
                            coordenadas_barco.append((i, y))
                    self.coordenadas_barco[nombre] = coordenadas_barco      #Asigno la lista al diccionario
                    colocado = True


    def validar_posicion(self, orientacion:str, x: int, y: int, eslora: int):
        ''' 
        Método para validar una posición de un barco.
        En función de la orientación del barco compruebo:
            1.- Primero que no salga del tablero.
            2.- Segundo que no se solape con otros barcos, es decir que la celda está vacía (agua).
        '''
        if orientacion == "E-O":
            if y + eslora > self.dimension:  
                return False
            for i in range(y, y + eslora):      
                if self.tablero_barcos[x][i] != SIMBOLO_AGUA:
                    return False
        else:       #Es decir orientacion == "N-S"
            if x + eslora > self.dimension:
                return False
            for i in range (x, x + eslora):
                if self.tablero_barcos[i][y] != SIMBOLO_AGUA:
                    return False
        return True     #Si supera todos los "filtros"


    def colocar_barco_en_tablero(self, orientacion:str, x: int, y: int, eslora: int):
        ''' 
        Métodos para colocar un barco en el tablero.

        '''
        if orientacion == "E-O":
            for i in range (y, y + eslora):
                self.tablero_barcos[x][i] = SIMBOLO_BARCO
        else:
            for i in range (x, x + eslora):
                self.tablero_barcos[i][y] = SIMBOLO_BARCO


    def mostrar_tablero(self):
        ''' 
        Método para mostrar el tablero de los barcos colocados en pantalla.
        '''
        print()
        print(f"******TABLERO DE BARCOS DE {self.jugador.upper()}******")
        pprint.pprint(self.tablero_barcos)
        print()
        print(f"*****TABLERO DE DISPAROS DE {self.jugador.upper()}*****")
        pprint.pprint(self.tablero_disparos)


    def disparo_coordenada(self, x: int, y: int):
        ''' 
        Método disparo coordenadas.Realizo varias comprobaciones:
            1.- Primero que la coordenada caiga dentro del tablero. Entero entre 0 y 9.
            2.- Segundo eue no haya disparado previamente a dicha posición.
            3.- Tercero si ha impactado a un barco.
            4.- Cuarto si ha impactado en agua.
        
        '''
        if not (0 <= x <= (self.dimension - 1) and 0 <= y <= (self.dimension - 1)):
            return "FIN", None
        if self.tablero_disparos[x][y] in [SIMBOLO_TOCADO, SIMBOLO_FALLO]:
            return "REPETIR", None
        
        if self.tablero_barcos[x][y] == SIMBOLO_BARCO:
            self.tablero_disparos[x][y] = SIMBOLO_TOCADO
            barco_afectado = None
            for barco, coordenadas in self.coordenadas_barco.items():
                if (x, y) in coordenadas:
                    coordenadas.remove((x, y))
                    barco_afectado = barco
                    break

            if barco_afectado and len(self.coordenadas_barco[barco_afectado]) == 0:
                return "HUNDIDO", barco_afectado
            
            else:
                return "TOCADO", None
        else:
            self.tablero_disparos[x][y] = SIMBOLO_FALLO
            return "AGUA", None


    def quedan_barcos(self):
        ''' 
        Método para comprobar si quedan barcos.
        Recorre el diccionario de barcos. 
        Mientras queden listas, el jugador no ha terminado con todos los barcos.
        '''
        for coordenadas in self.coordenadas_barco.values():
            if len(coordenadas) > 0:
                return True
        return False 


    def turno_maquina(tablero_jugador):
        while True:
            x = random.randint(0, DIMENSION - 1)
            y = random.randint(0, DIMENSION - 1)

            if tablero_jugador.tablero_barcos[x][y] == SIMBOLO_TOCADO or tablero_jugador.tablero_barcos[x][y] == SIMBOLO_FALLO:
                continue

            resultado, barco_afectado = tablero_jugador.disparo_coordenada(x,y)     
            return resultado, barco_afectado

    def solicitar_coordenada(self, elemento_coordenada: str):
        ''' 
        Método para pedir coordenada y comprobar que es un entero. x = fila, y = columna.
        Que este entre 0 y 9 (DIMENSION - 1) lo hace en otra parte del código.
        '''
        while True:
            try:
                valor = int(input(f"Introduzca {elemento_coordenada} (0-{self.dimension - 1})"))
                if 0 <= valor < self.dimension:
                    return valor
                print(f"La coordenada ({elemento_coordenada}) está fuera del tablero.")
            except ValueError:
                print("El valor introducido no es válido. Por favor, introduce un número entero.")


def iniciar_juego(): 
    ''' 
    Muestra el menú inicial y procesa la opción elegida por el usuario.
        Robustez:
            1.- En try/except compruebo que el usuario introduce un número entero (int)
            2.- Y con match/case si el número entero está entre 1 y 3
    '''
    while True:
        print (f"******BIENVENIDO A HUNDIR LA FLOTA******")
        try:
            opcion = int(input(f"Elige una opción. 1: Jugar, 2: Mostrar instruciones, 3: salir"))
        except ValueError:
            print(f"Por favor, introduzca un número entero válido (1: Jugar, 2: Mostrar instrucciones o 3: Salir)")
            continue

        match opcion:
            case 1:
                print()
                print(f"¡¡¡CORSARIO!!! Comienza la batalla...")
                return "JUGAR"

            case 2:
                print()
                print(f"Instrucciones:")
                print(f"Hay dos jugadores: tú y la máquina. Y un tablero de {DIMENSION}x{DIMENSION} posiciones donde irán los barcos")
                print(f"Lo primero que se hace es colocar los barcos de manera aleatoria.")
                print(f"4 barcos de 1 posición de eslora")
                print(f"3 barcos de 2 posiciones de eslora")
                print(f"2 barcos de 3 posiciones de eslora")
                print(f"1 barco de 4 posiciones de eslora")
                print(f"Tanto tú, como la máquina tenéis un tablero con barcos, y se trata de ir disparando y hundiendo los del adversario hasta que un jugador se queda sin barcos, y por tanto, pierde.")
                print(f"Funciona por turnos y empiezas tú.")
                print(f"En cada turno disparas a una coordenada (x,y) del tablero adversario. **Si aciertas, te vuelve a tocar**. En caso contrario, le toca a la máquina.")
                print(f"En los turnos de la máquina, si acierta también le vuelve a tocar.")
                print(f"Si se hunden todos los barcos de un jugador, el juego acaba y gana el otro.")
                print()

            case 3:
                print()
                print(f"Parece que hoy no tienes la energía suficiente :P. ¡¡¡LASTER ARTE!!!")
                return "SALIR"

            case _:
                (f"Opción incorrecta, introduzca un número entero válido (1: Jugar, 2: Mostrar instrucciones o 3: Salir)")

def jugar():
#Muestra el menú inicial def en la fn "iniciar_juego" uy procesa la opción elegida
    decision = iniciar_juego()
    if decision == "SALIR":
        return

#Si la opción elegida NO es "SALIR" crea los tableros
    nombre_usuario = input(f"Introduce tu nombre, corsario. En caso contrario, pulsa enter.")
    if not nombre_usuario:   
        nombre_usuario = "Usuario"
    
    tablero_humano = Tablero(jugador=nombre_usuario)
    tablero_maquina = Tablero(jugador="Máquina")


#Completa ambos tableros de manera alatoria con la fn "colocar_barcos_random"
    tablero_humano.colocar_barcos_random()
    tablero_maquina.colocar_barcos_random()

    print(f"Barcos colocados de manera aleatoria. ¡¡¡comienza la partida!!! GOOD LUCK!!!")
    tablero_humano.mostrar_tablero()

#Comienza la partida. COMIENZA EL USUARIO!!!
    while True:   
        turno_jugador = True
        
        while turno_jugador:
            print(f"Tu turno, {nombre_usuario}")
    #Solicita las coordenadas
            x = tablero_humano.solicitar_coordenada("Fila (x)")
            y = tablero_humano.solicitar_coordenada("Columna (y)")

    #Dispara sobre el tablero
            resultado = tablero_maquina.disparo_coordenada(x,y)

            match resultado:
                case "TOCADO":
                    print (f"¡¡¡TOCADO!!! Sigues jugando")

                case "HUNDIDO":
                    print (f"¡¡¡TOCADO y HUNDIDO!!!. Has hundido: {barco_afectado}")
                    if not tablero_maquina.quedan_barcos():
                        print(f"ZORIONAK!!! Con este, has hundido todos los barcos de tu adversario. YOU WIN!!!")
                        return      #Rompe el bucle while y se acaba la partida

                case "AGUA":
                    print(f"¡¡¡AGUA!!! Pierdes tu turno...")
                    turno_jugador = False

                case "REPETIR":
                    print(f"Ya habías disparado en esa posición, por favor, elige otra coordenada")
                    continue        #Bucle while, salta arriba y vuelve a empezar        

                case "FIN":
                    print(f"La coordenada elegida está fuera del tablero.")
                    continue        #Bucle while, salta arriba y vuelve a empezar

    #Turno de la MÁQUINA
        turno_maquina = True

        while turno_maquina:
            print(f"Turno de la máquina.")

            resultado = turno_maquina(tablero_humano)

            match resultado:
                case "TOCADO":
                    print (f"¡¡¡TOCADO!!!")

                case "HUNDIDO":
                    print (f"¡¡¡TOCADO y HUNDIDO!!! La máquina ha hundido {barco_afectado}")
                    if not tablero_humano.quedan_barcos():
                        print(f"Con este, la máquina ha hundido todos tus barcos. YOU LOSE!!!")
                        return      #Rompe el bucle while y se acaba la partida

                case "AGUA":
                    print(f"¡¡¡AGUA!!!")
                    turno_maquina = False

                case "REPETIR":
                    continue        #Bucle while, salta arriba y vuelve a empezar        

                case "FIN":
                    continue        #Bucle while, salta arriba y vuelve a empezar


        if __name__ == "__main__":
            jugar()
```
