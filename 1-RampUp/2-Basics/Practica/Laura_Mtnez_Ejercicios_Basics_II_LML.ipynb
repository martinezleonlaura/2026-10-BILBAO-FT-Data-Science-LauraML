## Ejercicio 4
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C 
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
A = 4
B = "Text"
C = 4.1

if A == B:
    print("A y B son iguales")
else:
    print("A y B NO son iguales")

if A != C:
    print("A y C NO son equivalentes")
else:
    print("A y C son equivalentes")

if A > C:
    print("A es mayor que C")
else:
    print("A NO es mayor que C")

if C <= A:
    print("C es menor o igual que A")
else:
    print("C NO es menor o igual que A")

if B != C:
    print("B y C NO son equivalentes")
else:
    print("B y C son equivalentes")
````

## Ejercicio 7
Crea un programa que recoja dos inputs. Tiene que comprobar si su suma es igual, superior o inferior a 10, cada condición por separado

````python
num_1 = int(input("Escribe un número: "))
num_2 = int(input("Escribe otro número: "))
suma = num_1 + num_2
cond_1 = (suma == 10)
cond_2 = (suma > 10)
cond_3 = (suma < 10)
print (f"La suma de {num_1} y {num_2} es {suma}")
print (f"La suma de {num_1} y {num_2} es {suma} es igual a 10: {cond_1}")
print (f"La suma de {num_1} y {num_2} es {suma} es mayor que 10: {cond_2}")
print (f"La suma de {num_1} y {num_2} es {suma} es menor que 10: {cond_3}")
````

## Ejercicio 10
