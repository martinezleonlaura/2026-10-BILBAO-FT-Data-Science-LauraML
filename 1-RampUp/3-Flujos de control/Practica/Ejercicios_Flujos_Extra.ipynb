## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
operaciones_realizadas = []

while True:         #Para que siga funcionando hasta que el usuario elija salir
    operacion = input("Del 1 al 6 ¿Qué operación quieres realizar? 1:multiplicación, 2:división, 3:suma, 4:resta, 5:potencia, 6:salir:")
    try:            #Para que sea robusta frente a entradas que no sean num enteros
        operacion = int(operacion)
    except ValueError:
        print("Error, introduzca un número válido.")
        continue

    if operacion == 6:          #Opción salir, pregunta si quiero imprimir el historial
        historial = input("Has elegido salir, ¿quieres mostrar el historial de ops?: 1:sí, 2:no")
        if historial == "1":
            print ("El historial de operaciones es el siguiente: ")
            if not operaciones_realizadas:      #Si la lista del historial está vacía
                print("No se realizaron operaciones")
            else:
                for op in operaciones_realizadas:
                    print(op)
        print("Has elegido salir, ¡Hasta pronto!")
        break       #Para terminar el programa

    if operacion < 1 or operacion > 5:          #Una vez comprobado que es un numero entero, compruebo que el num está entre 1 y 5 (operaciones)
        print ("Error, introduzca un número válido (entre 1 y 5)")
        continue

    try:            #Pedimos los números para operar
        num_1 = float(input("Introduce el primer número: "))
        num_2 = float(input("Introduce el segundo número: "))
    except ValueError:
        print("Error, introduzca un número real válido.")
        continue
    
    match operacion:
        case 1:
            mult = num_1 * num_2
            comentario = (f"{num_1} x {num_2} = {mult}")
            print (comentario)
            operaciones_realizadas.append(comentario)
        case 2:
            if num_2 == 0:
                print ("Error: el resultado de dividir entre cero es infinito")
            else:
                div = num_1 / num_2
                comentario = (f"{num_1} / {num_2} = {div}")
                print (comentario)
                operaciones_realizadas.append(comentario)
        case 3:
            suma = num_1 + num_2
            comentario = (f"{num_1} + {num_2} = {suma}")
            print (comentario)
            operaciones_realizadas.append(comentario)
        case 4:
            resta = num_1 - num_2
            comentario = (f"{num_1} - {num_2} = {resta}")
            print (comentario)
            operaciones_realizadas.append(comentario)
        case 5:
            pot = num_1 ** num_2
            comentario = (f"{num_1} ^ {num_2} = {pot}")
            print (comentario)
            operaciones_realizadas.append(comentario)
        case _:
            print ("Error, introduzca un número válido (número entre 1 y 6)")
```
