Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:

* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True

    * Nombre - String

```python
#Implemento la clase Tienda
class Tienda:

#Atributos COMUNES:
    tipo = "electrodomésticos"
    abierta = True

#Atributos PROPIOS de cada Tienda:
    def __init__(self, nombre:str, direccion:str, num_empleados:int, ventas_ult_3_meses:list):
        self.nombre = nombre
        self.direccion = direccion
        self.num_empleados = num_empleados
        self.ventas_ult_3_meses = ventas_ult_3_meses

#Método para calcular las ventas totales en los últimos 3 meses
    def ventas_totales(self):
        return sum(self.ventas_ult_3_meses)
    
#Método que calcula la media de ventas de los ultimos meses, por empleado
    def media_ventas_x_empleado(self):
        return self.ventas_totales() / self.num_empleados
    
#Método que devuelve en un string el nombre de la tienda, junto con su dirección.
    def nombre_direccion(self):
        return str (self.nombre, "_", self.direccion)

#Método para obtener las ventas del último mes de la lista de ventas de los 3 últimos meses
    def ventas_ultimo_mes(self):
        return self.ventas_ult_3_meses[-1]

#Método para obtener las ventas del último mes
    def ventas_ultimo_mes(self):
        return self.ventas_ult_3_meses[-1]

#Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.
    def proyeccion_ventas(self):
        if X < 1000:
            return self.ventas_ult_3_meses * 1.2
        else:
            return self.ventas_ult_3_meses * 1.5

#Creo 3 tiendas
tienda_1 = Tienda("Primera", "Avenida Buenos Aires", 4, [10000, 10000, 10000])
tienda_2 = Tienda("Segunda", "Gran Vía", 5, [20000, 20000,20000])   
tienda_3 = Tienda("Tercera", "Avenida Colón", 2, [5000, 5000, 10000])

#Compruebo en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
print("***")
print("Tipo:", tienda_1.tipo)
print("Abierta:", tienda_1.abierta)
print("Nombre:", tienda_1.nombre)
print("Dirección:", tienda_1.direccion)
print("Número de empleados:", tienda_1.num_empleados)
print("Ventas últimos 3 meses:", tienda_1.ventas_totales())
print("Media de ventas por empleado en el mes:", tienda_1.media_ventas_x_empleado())
print("***")

#Calculo las ventas del último mes de todas las tiendas usando buble for
print("***")
tiendas = [tienda_1, tienda_2, tienda_3]
for tienda in tiendas:
    print ("Las ventas del último mes de", tienda.nombre, ":", tienda.ventas_ultimo_mes())
print("***")

#Imprimo por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
print("***")
for tienda in tiendas:
    if "Avenida" in tienda.direccion:
        print (f"Tienda: {tienda.nombre}, Dirección: {tienda.direccion}")
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
#Implemento la clase Perro
class Perro:

#Atributos COMUNES:
    patas = 4
    orejas = 2
    ojos = 2

#Atributos PROPIOS de cada Perro:
    def __init__(self, raza, pelo = "marrón", duegno = False):
        self.raza = raza
        self.pelo = pelo
        self.duegno = duegno
        self.velocidad = 0

#Diferentes métodos para la Clase Perro:
    def andar(self, aumento_velocidad):
        self.velocidad = self.velocidad + aumento_velocidad

    def parar(self):
        self.velocidad = 0

    def ladrar(self, ladrido):
        return str(ladrido) + " GUAU! "

#Creo un objeto tipo Perro, sin dueño
perro_1 = Perro (raza = "caniche")
print(type(perro_1))

#Compruebo que están bien todos sus atributos
print ("patas:", perro_1.patas, ", orejas:", perro_1.orejas, ", ojos:", perro_1.ojos, ", raza:", perro_1.raza, ", pelo:", perro_1.pelo, ", dueño:", perro_1.duegno, ", velocidad:", perro_1.velocidad)

#Compruebo que anda, le doy velocidad
perro_1.andar(9)
print("Compruebo que el perro tiene una velocidad de:", perro_1.velocidad)

#Paro el perro, velocidad = 0
perro_1.parar()
print("Compruebo que el perro está parado, por lo que tiene una velocidad de:", perro_1.velocidad)

#Hago que el perro ladre
perro_1.ladrar("¡Ladra!")
```
