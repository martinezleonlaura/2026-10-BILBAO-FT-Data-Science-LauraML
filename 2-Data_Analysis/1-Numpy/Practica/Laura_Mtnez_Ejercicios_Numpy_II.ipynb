## Ejercicio 12
Encuentra los índices de los valores que no sean 0 del siguiente vector [1,2,0,0,4,0]. El output sería [1, 2 4]. [Encuentra el método apropiado en la documentación.](https://numpy.org/doc/) Encapsúlalo en una función, con un argumento de entrada de un vector.

````python
array_ej_12 = np.array([1, 2, 0, 0, 4, 0])
print(array_ej_12)

def encontrar_valores(vector):
    return np.nonzero(vector)[0]        #Nonzero devulve una tupla, entonces pongo [0] para que me ponga el primer elemento

print(encontrar_valores(array_ej_12))
````

## Ejercicio 18
Crea una matriz aleatoria de 5x5 y normaliza sus valores. Para normalizar tienes que restar sus valores por la media(`np.mean`) y dividir el resultado por la desviación estándar(`np.std`).

````python
array_ej_18 = np.random.randint (1, 30, size = (5,5))
print(array_ej_18)

print()

array_ej_18 = (array_ej_18 - np.mean(array_ej_18)) / np.std(array_ej_18)
print(array_ej_18)
````

## Ejercicio 19
1. Crea una secuencia de números que vayan del 0 al 10.
2. Crea un nuevo array, que parta del anterior, pero sustituya por el valor -1, todos los elementos cuyo valor se encuentre entre el 3 y el 8.
3. Crea un nuevo array que niegue todos los elementos entre el 3 y el 8

````python
#Apt.1
array_ej_19 = np.arange(10)
print(array_ej_19)
print()

#Apt.2
array_ej_19_new = array_ej_19.copy()
condicion = (array_ej_19_new >= 3) & (array_ej_19_new <= 8)
array_ej_19_new = np.where(condicion, -1, array_ej_19_new)
print(array_ej_19_new)
print()

#Apt.3
array_ej_19_new_2 = array_ej_19.copy()
condicion = (array_ej_19_new_2 >= 3) & (array_ej_19_new_2 <= 8)
array_ej_19_new_2 = np.where(condicion, array_ej_19_new_2 * -1, array_ej_19_new)
print(array_ej_19_new_2)

````
