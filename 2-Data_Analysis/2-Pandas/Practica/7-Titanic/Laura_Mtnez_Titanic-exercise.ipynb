9. What is the name of the oldest person who died in the Titanic? Was he or she travelling alone or had any family travelling with them?  



In order to give an answer to the second question you should find out which columns give you that info. Usually part of your job as a Data Scientist will be get to know the dataset which you are working with. In this case the columns which give you that info are the following: 
        - 'sibsp' Number of Siblings/Spouses Aboard
        - 'parch' Number of Parents/Children Aboard

```python
#Primero saco los fallecidos
deads = titanic[titanic['survived'] == 0]

#Segundo busco el de mayor edad idxmax
oldest_person_who_died = deads.loc[deads['age'].idxmax()]
print(f"The oldest person who died was {oldest_person_who_died['name']} at the age of {oldest_person_who_died['age']}")
```

```python
#Ahora veo si viajaba acompañado
print(f"Was {oldest_person_who_died['name']} travelling alone?: {oldest_person_who_died['sibsp']}")
print(f"Had {oldest_person_who_died['name']} any family travelling with him?: {oldest_person_who_died['parch']}")
```

```python
#Voy a comprobar si había más pasajeros con la misma edad
deads_without_nan = deads.dropna(subset=['age'])  #Quito los nan
deads_sorted = deads_without_nan.sort_values(by='age')   #Ordeno por edad, de menos a más
deads_sorted.tail(2)   #saco lo 2 últimos para ver si 74.0 es la mayor edad --> está OK
```

24. Using [`.pivot_table()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.pivot_table.html) to count how many women or men survived by class, call it `counted`.

Don't panic and read the documentation!




```python
counted = titanic.pivot_table(values='survived', index='pclass', columns='gender', aggfunc='sum')
counted
```

34. Using [`kdeplot`]("https://seaborn.pydata.org/generated/seaborn.kdeplot.html") from `seaborn` represent those who not survived distributed by age.<>

Hint: First you should "filter" the `titanic` dataset where the column "survived" is 0, indexing the column `"age"` only.

Arguments you should pass to the function: 

    - color = "red"
    - label = "Not Survived"
    - shade = True
    
You should get something like this: 

![kdeplot](./img/kdeplotsur.png)

```python
#Primero filtro los que no sobrevivieron y me quedo con la columna edad
not_survived_age = titanic[titanic['survived'] == 0]['age']

sns.kdeplot(not_survived_age, color='red', label='Not Survived', fill=True)   #Shade está obsoleto, ahora es fill
plt.title('Age Distribution of Those Who Did Not Survive')
plt.xlabel('Age')
plt.legend()
plt.show()
```
