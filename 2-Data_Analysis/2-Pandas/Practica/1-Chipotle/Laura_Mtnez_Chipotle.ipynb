# Ex1 - Filtering and Sorting Data



### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
chipo_new = chipo.copy()
chipo_new['item_price'] = chipo_new['item_price'].str.replace('$', '',)
chipo_new['item_price'] = chipo_new['item_price'].astype(float)
chipo_new['precio_ud'] = chipo_new['item_price'] / chipo_new['quantity']

producto_mas_caro = chipo_new.loc[chipo_new['precio_ud'].idxmax(), 'item_name']
print(f"El productos más caro es: {producto_mas_caro}")
```

### Step 4. How many products cost more than $10.00?

```python
productos_mas_10 = chipo_new[chipo_new['precio_ud'] > 10]
total_productos = productos_mas_10['item_name'].nunique()
print(f"Hay {total_productos} productos que valen más de 10$.")
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
total_pedidos = productos_mas_10['order_id'].nunique()
print(f"Sehan hecho {total_pedidos} pedidos con un productos de más de $10.")
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
#Agrupamos por pedido y sumamos la columna item_price
#Estos nos da una lista con lo que gastó cada order_id en total
totales_por_pedido = chipo_new.groupby('order_id')['item_price'].sum()
pedidos_mas_10 = (totales_por_pedido > 10).sum()
print(f"Se han hecho {pedidos_mas_10} pedidos cuyo importe total supera los $10.")
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
#Cuento los pedidos únicos en los que se da esta condición
filas_mas_10 = chipo_new[chipo_new['item_price'] > 10]
total_pedidos_mas_10_produc = filas_mas_10['order_id'].nunique()
print(f"En {total_pedidos_mas_10_produc} pedidos se pagó más de $10 por un mismo producto.")
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
resultado = chipo_new.drop_duplicates(subset=['item_name', 'precio_ud'])[['item_name', 'precio_ud']]
print(resultado)
```

### Step 6. Sort by the name of the item

```python
resultado_2 = resultado.sort_values(by='item_name').reset_index(drop=True)
print(resultado_2)
```

### Step 7. What was the quantity of the most expensive item ordered? 2 ways

```python
#Filtrando por el nombre del producto más caro
nombre_mas_caro = chipo_new.loc[chipo_new['precio_ud'].idxmax(), 'item_name']
cantidad_f1 = chipo_new[chipo_new['item_name'] == nombre_mas_caro]['quantity'].sum()
print(f"Cantidad ordenada: {cantidad_f1}")
```

```python
#Agrupando por producto y buscando el máximo
cantidades_por_producto = chipo_new.groupby('item_name')['quantity'].sum()
cantidad_f2 = cantidades_por_producto[nombre_mas_caro]
print(f"Cantidad ordenada: {cantidad_f2}")
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad_count = len(chipo_new[chipo_new['item_name'] == "Veggie Salad Bowl"])
print(f"Se pidieron {veggie_salad_count} Veggie Salad Bowls.")
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
filtro_canned_soda = chipo_new[(chipo_new['item_name'] == 'Canned Soda') & (chipo_new['quantity'] > 1)]
veces_pedido = filtro_canned_soda.shape[0]   #0 para contar filas
print(f"Alguien pidió más de una Canned Soda {veces_pedido} veces.")
```
