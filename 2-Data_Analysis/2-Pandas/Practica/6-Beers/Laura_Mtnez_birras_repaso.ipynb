## Ejercicio 23
Consigue en un DataFrame de todas las cervezas en cuyas cervecerías al menos haya una cerveza con un `abv` mínimo 0.07.

```python
#Saco las cervecerías que tienen alguna cerveza con abv >= 0.07
breweries_validas = combinado.loc[combinado['abv'] >= 0.07, 'brewery_id'].unique()
```

```python
#Filtro el DataFrame original para quedarnos con todas las cervezas de esas cervecerías
beers_filtradas = combinado[combinado['brewery_id'].isin(breweries_validas)].sort_values(by='abv', ascending=False)
```

## Ejercicio 24
Teniendo en cuenta la media de amargor de todas las cervezas, obtén una tabla con las cervezas cuyos estados tengan una media de amargor en cerveza por encima de la media

```python
#media de amargor de TODAS las cervezas
media_amargor_global = combinado['ibu'].mean().round(2)
print(media_amargor_global)
```

```python
#calculo la media de amargor por cada estado
media_amargor_estado = combinado.groupby('state')['ibu'].mean()
#media_amargor_estado
```

```python
#filtro los estados que superan la media global
estados_que_cumplen = media_amargor_estado[media_amargor_estado > media_amargor_global].index
```

```python
#me quedo solo con las cervezas que pertenecen a esos estados
combinado[combinado['state'].isin(estados_que_cumplen)].sort_values(by='ibu', ascending=False)
```
