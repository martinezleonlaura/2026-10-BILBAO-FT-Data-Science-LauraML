El objetivo es ordenar la típica carpeta de cajón desastre donde se nos acumulan archivos de
todos los formatos como: imágenes, words, presentaciones, excels, ejecutables, zips, etc... Y
que no vamos a ir viendo archivo a archivo, ordenándolos.... La solución suele ser eliminarlo
todo. Vamos a intentar no tener que llegar a eso, ordenando los ficheros automáticamente en
carpetas y revisar las que nos interesen. La típica carpeta donde aplicaríamos este programa es
la de “Descargas”, donde se suelen acumular muchos archivos temporales.


Por tanto, **el programa tiene que guardar los archivos de la carpeta “Descargas” (o cualquier
otra donde acumules mucho archivo temporal) en los siguientes directorios dentro de la
propia carpeta “Descargas”**:
- Imagenes
- Documentos
- Software
- Otros

Cada vez que ejecutes el script, automáticamente se ordenarán todos los archivos en sus
correspondientes carpetas.


```python
doc_types = ('.doc', '.docx', '.txt', '.pdf', '.xls', '.ppt', '.xlsx', '.pptx')
img_types = ('.jpg', '.jpeg', '.png', '.svg', '.gif')
software_types = ('.exe', '.py','.ipynb')
```

### 1. Crea un un programa con el flujo principal y encapsúlalo en funciones.

```python
import os
import shutil
```

```python
def ordenar_archivos(ruta):

    #Guardo la ruta inicial, para después volver a ella
    ruta_inicial = os.getcwd()

    #Voy a la ruta que quiero oredenar
    os.chdir(ruta)

    #Creo las carpetas donde iré guardando los archivos en fn del tipo
    #Hago un Try/Except para cuando estén ya las carpetas creadas no me de error
    carpetas_destino = {
        'Imagenes': img_types,
        'Documentos': doc_types,
        'Software': software_types,
        'Otros': () # Para cualquier otra extensión
    }
    
    for carpeta in carpetas_destino.keys():
        try:
            os.mkdir(carpeta)
        except FileExistsError:
            pass   #Si ya existe, simplemente continúa con la siguiente sin romperse

    lista_archivos = os.listdir() #Lo saco fuera para que no cambie en mitad del bucle

    #Recorro el diccionario
    for nombre_carpeta, extensiones in carpetas_destino.items():
        # Nos saltamos 'Otros' de momento, lo gestionamos al final
        if nombre_carpeta == 'Otros':
            continue

        #recorro las listas
        for extension in extensiones:
            for archivo in lista_archivos:
                if archivo.endswith(extension):
                    shutil.move(archivo, f"{nombre_carpeta}/")

    #Reviso la lista con lo que queda
    for archivo in os.listdir():
        if archivo not in carpetas_destino.keys():
            shutil.move(archivo, 'Otros/')


    resultado = os.walk(os.getcwd())

    #vuelvo a la ruta inicial 
    os.chdir(ruta_inicial)

    return [x for x in resultado]

```

```python
ruta = 'C:\\Users\\marti\\Desktop\\Descargas'
```

```python
ordenar_archivos(ruta)
```
