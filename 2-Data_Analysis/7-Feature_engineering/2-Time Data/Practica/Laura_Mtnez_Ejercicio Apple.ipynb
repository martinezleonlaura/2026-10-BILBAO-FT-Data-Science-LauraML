# Apple Stock

### Introduction:

We are going to use Apple's stock price.


### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
from datetime import datetime
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/09_Time_Series/Apple_Stock/appl_1980_2014.csv)

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/09_Time_Series/Apple_Stock/appl_1980_2014.csv'
```

### Step 3. Assign it to a variable apple and apply a head()

```python
apple = pd.read_csv(url, sep=",")
```

```python
apple.head()
```

### Step 4.  Check out the type of the columns

```python
apple.dtypes
```

### Step 5. Transform the Date column as a datetime type and check the info()

```python
apple['Date'] = pd.to_datetime(apple['Date'])
```

```python
apple.info()
```

```python
apple.head()
```

### Step 6.  Set the date as the index

```python
apple = apple.set_index('Date')
```

```python
apple.head()
```

### Step 7.  Is there any duplicate dates?

```python
duplicados = apple[apple.index.duplicated(keep=False)]
```

```python
duplicados
```

```python
#No hay duplicados
```

### Step 8.  Ops...it seems the index is from the most recent date. Make the first entry the oldest date.

```python
apple = apple.sort_index(ascending=True)
```

```python
apple.head()
```

### Step 9. Get the last business day of each month

```python
apple.resample('BME').last()
```

### Step 10.  What is the difference in days between the first day and the oldest

```python
diferencia_dias = (apple.index.max() - apple.index.min()).days
print(diferencia_dias)
```

### Step 11.  How many months in the data we have?

```python
total_meses = apple.index.to_period('M').nunique()
print(total_meses)
```

### Step 12. Plot the 'Adj Close' value. Set the size of the figure to 13.5 x 9 inches

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(13.5,9))

plt.plot(apple.index, apple['Adj Close'])

plt.title("Evolución de Adj Close")
plt.xlabel("Fecha")
plt.ylabel("Adj Close")

plt.show()
```
