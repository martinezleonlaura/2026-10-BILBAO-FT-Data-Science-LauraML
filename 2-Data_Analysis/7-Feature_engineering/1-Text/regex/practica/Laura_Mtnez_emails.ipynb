# Extrae los dominios de los correos con una expresión regular

```python
import re

emails = '''
miguellc@thebridgeschool.tech
miguel.cruz@uc3m.es
miguel-12-2000@mi-correo.net
miguellc@s@thebridgeschool.es
'''
```

Ejemplo:
* @thebridgeschool.tech
* @uc3m.es
* @mi-correo.net
* @thebridgeschool.es

```python
pattern = re.compile(r'@[a-zA-Z0-9.-]+\.\w*')  #'r' para que me interprete literalmente

matches = pattern.finditer(emails)

for match in matches:
    print(match)
```
